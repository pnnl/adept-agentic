# Multi-Agent Collaboration

This notebook demonstrates the ADEPT framework's multi-agent orchestration capabilities.

## Security Architecture

All API requests route through `nginx_proxy` (TLS + JWT authentication).
Multi-agent sessions are managed server-side by the orchestration service.

## Features Covered
- Creating multi-agent sessions
- Task distribution and coordination
- Agent communication patterns
- Parallel task execution
- Result aggregation

## Docker Configuration
- JupyterLab service: `deployment/local/docker-compose/docker-compose.core.yaml`
- Rebuild after config changes: `make rebuild-jupyterlab`

```python
import os
import json
import httpx
import openai
from datetime import datetime

# Load authentication
auth = json.load(open("/tmp/adept_auth.json"))
config = json.load(open("/tmp/adept_config.json"))

API_BASE_URL = config.get('api_base_url', os.getenv('ADEPT_API_BASE_URL', 'https://nginx_proxy/v1'))

client = openai.OpenAI(
    base_url=API_BASE_URL,
    api_key=auth['access_token'],
    http_client=httpx.Client(verify=False)
)

print("Multi-Agent Environment Ready")
print(f"   API: {API_BASE_URL}")
print(f"   Routing: nginx_proxy (TLS + JWT)")
```

## Multi-Agent Task Example

```python
# Complex task requiring multiple agentstask = '''I need help with a comprehensive research project:1. Search for recent developments in quantum computing2. Analyze the data and create visualizations3. Generate a summary report with recommendationsPlease coordinate multiple agents to complete this efficiently.'''response = client.chat.completions.create(    model="agentic-framework:scientific-agent-v1",    messages=[{"role": "user", "content": task}],    temperature=0.7,    max_tokens=2000)print(response.choices[0].message.content)
```

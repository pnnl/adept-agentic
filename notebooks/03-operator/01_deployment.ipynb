## Kubernetes Deployment

```python
# Helm values example
helm_values = '''
global:
  environment: production
  domain: api.agentic.local

agent_gateway:
  replicas: 3
  resources:
    requests:
      memory: "1Gi"
      cpu: "500m"
    limits:
      memory: "2Gi"
      cpu: "1000m"

orchestration_service:
  replicas: 2
  autoscaling:
    enabled: true
    minReplicas: 2
    maxReplicas: 10
    targetCPU: 70

postgresql:
  persistence:
    size: 50Gi
  backup:
    enabled: true
    schedule: "0 2 * * *"
'''

print(helm_values)
```
